```python
import pandas as pd
import numpy as np
pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("TR8_db_extracted_traits.csv")
df.head(20)
```

```python
# df[["scientific_name", "ECO_Height_Max", "ECO_Height_Min", "LEDA_Lifespan"]]
df["scientific_name"] = df["scientific_name"].str.lower()

df["ECO_Height_Max"] = pd.to_numeric(df["ECO_Height_Max"], errors="coerce")
df["ECO_Height_Min"] = pd.to_numeric(df["ECO_Height_Min"], errors="coerce")

# height
df["ECO_Height_Mean"] = df[["ECO_Height_Max", "ECO_Height_Min"]].mean(axis=1)
df_height = (
    df.groupby("scientific_name", as_index=False)
      .agg(
          height_max=("ECO_Height_Max", "max"),
          height_min=("ECO_Height_Min", "min"),
          height_mean=("ECO_Height_Mean", "mean")
      )
)


df_height
```

```python
df_height.to_csv("TR8_db_df_height.csv", index=False)
```

```python
def mode_or_nan(s):
    m = s.dropna().mode()
    return m.iloc[0] if not m.empty else pd.NA

df_lifespan = (
    df.groupby("scientific_name", as_index=False)
      .agg(
          lifespan=("LEDA_Lifespan", mode_or_nan)
      )
)
df_lifespan
```

```python
df[df["LEDA_Lifespan"]=="annuals + perennials + summer annuals + winter annuals + strict monocarpic bi-annuals and poly-annuals"]
```

```python
df = df.copy()

df.loc[:, "LEDA_Lifespan_clean"] = (
    df["LEDA_Lifespan"]
    .str.extract(r"\(([^)]+)\)", expand=False)
)

lifespan_map = {
    "perennials": "perennial",
    "summer annuals": "annual",
    "annuals + summer annuals": "annual",
    "summer annuals + strict monocarpic bi-annuals and poly-annuals": "annual",
    "annuals": "annual"
}

df.loc[df["LEDA_Lifespan_clean"].isna(), "LEDA_Lifespan_clean"] = (
    df.loc[df["LEDA_Lifespan_clean"].isna(), "LEDA_Lifespan"]
      .map(lifespan_map)
)
df_check = df[df["LEDA_Lifespan"].notna()]
df_check = df_check[["scientific_name", "LEDA_Lifespan_clean"]]
df_check
```
