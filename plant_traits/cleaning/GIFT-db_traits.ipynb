```python
import pandas as pd
import numpy as np
pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("GIFT-db_traits.csv")
df.head(20)
```

```python
# plant height
df_plant_height = df[[
    "work_species", 
    "trait_value_Plant_height_min",
    "trait_value_Plant_height_max",
    "trait_value_Plant_height_mean",
    "work_author"
]].copy()

df_plant_height["work_species"] = (df_plant_height["work_species"].astype(str).str.strip().str.lower())
df_plant_height = df_plant_height.dropna(subset=["trait_value_Plant_height_min", 
                                                 "trait_value_Plant_height_max",
                                                 "trait_value_Plant_height_mean"],
                                         how="all")

df_plant_height = df_plant_height.reset_index(drop=True)

df_plant_height
```

```python
# SSD
df_ssd = df[[
    "work_species",
    "trait_value_SSD_mean",
    "work_author"
]].copy()

df_ssd["work_species"] = (df_ssd["work_species"].astype(str).str.strip().str.lower())
df_ssd = df_ssd.dropna(subset=["trait_value_SSD_mean"])

df_ssd = df_ssd.reset_index(drop=True)

df_ssd
```

```python
# chromosone number
df_chrom = df[[
    "work_species", 
    "trait_value_Chromosome_number",
    "work_author"
]].copy()

df_chrom["work_species"] = (df_chrom["work_species"].astype(str).str.strip().str.lower())
df_chrom = df_chrom.dropna(subset=["trait_value_Chromosome_number"])

df_chrom = df_chrom.reset_index(drop=True)

df_chrom
```

```python
# lifecycle
df_l = df[[
    "work_species", 
    "trait_value_Lifecycle_1",
    "work_author"
]].copy()

df_l["work_species"] = (df_l["work_species"].astype(str).str.strip().str.lower())
df_l = df_l.dropna(subset=["trait_value_Lifecycle_1"])

df_l = df_l.reset_index(drop=True)

df_l
```

```python
# seed length
df_sl = df[[
    "work_species", 
    "trait_value_Seed_length_max",
    "trait_value_Seed_length_mean",
    "trait_value_Seed_length_min",
    "work_author"
]].copy()

df_sl["work_species"] = (df_sl["work_species"].astype(str).str.strip().str.lower())
df_sl = df_sl.dropna(subset=["trait_value_Seed_length_max", 
                            "trait_value_Seed_length_mean",
                            "trait_value_Seed_length_min"],
                    how="all")

df_sl = df_sl.reset_index(drop=True)

df_sl
```

```python
# reproduction sexual
df_rs = df[[
    "work_species",
    "trait_value_Reproduction_sexual_1",
    "work_author"
]].copy()

df_rs["work_species"] = (df_rs["work_species"].astype(str).str.strip().str.lower())
df_rs = df_rs.dropna(subset=["trait_value_Reproduction_sexual_1"])

df_rs = df_rs.reset_index(drop=True)

df_rs
```

```python
# leaf size
df_ls = df[[
    "work_species",
    "trait_value_Leaf_size_mean", 
    "work_author"
]].copy()

df_ls["work_species"] = (df_ls["work_species"].astype(str).str.strip().str.lower())
df_ls = df_ls.dropna(subset=["trait_value_Leaf_size_mean"])

df_ls = df_ls.reset_index(drop=True)

df_ls
```

```python
df_ls.to_csv("GIFT-db_leaf_size.csv", index=False)
df_sl.to_csv("GIFT-db_seed_length.csv", index=False)
df_plant_height.to_csv("GIFT-db_plant_height.csv", index=False)
df_ssd.to_csv("GIFT-db_stem_wood_density.csv", index=False)
df_chrom.to_csv("GIFT-db_chromosome_number.csv", index=False)
df_l.to_csv("GIFT-db_lifecycle.csv", index=False)
df_rs.to_csv("GIFT-db_reproduction_sexual.csv", index=False)
```
