```python
import pandas as pd
import numpy as np
pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("ssd.txt", sep='\t', encoding='latin-1')
df.head(20)
```

```python
df.groupby("AccSpeciesName").size()
```

```python
df = df[df["TraitID"].notna()]
df.head(20)
```

```python
df_trait_id = df[["AccSpeciesName", "StdValue", "UnitName", "Reference"]]
```

```python
df_trait_id_non_nan = df_trait_id[df_trait_id['StdValue'].notna()]

df_trait_id_non_nan.loc[:, "AccSpeciesName"] = (
    df_trait_id_non_nan["AccSpeciesName"].str.lower()
)

df_grouped = df_trait_id_non_nan.groupby("AccSpeciesName", as_index=False).agg(
    ssd_mean_g_cm3=("StdValue", "mean"),
    ssd_max_g_cm3=("StdValue", "max"),
    ssd_min_g_cm3=("StdValue", "min"),
    references=("Reference", lambda x: "; ".join(x.dropna().unique()))
)
df_grouped
```

```python
df_grouped.to_csv("ssd_cleaned.csv", index=False)
```
